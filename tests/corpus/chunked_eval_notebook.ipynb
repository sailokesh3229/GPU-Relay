# Chunked evaluation notebook (no framework, unmodified)

```python
import json
import math
import os
import random
import time

from IPython.display import display
```

```python
N_ITEMS = int(os.environ.get("CORPUS_ITEMS", "400"))
CHUNK_SLEEP = float(os.environ.get("CORPUS_CHUNK_SLEEP", "0"))
random.seed(1234)
items = list(range(N_ITEMS))
results = []
```

```python
def expensive(x):
    return math.sin(x) * random.random()
```

```python
for start in range(0, N_ITEMS, 20):
    values = [expensive(x) for x in items[start : start + 20]]
    results.append(values)
    print(f"chunk {start // 20:3d} mean {sum(values) / len(values):+.6f}")
    if CHUNK_SLEEP:
        time.sleep(CHUNK_SLEEP)
```

```python
flat = [v for chunk in results for v in chunk]
summary = {"n": len(flat), "mean": round(sum(flat) / len(flat), 10)}
display(summary)
summary["mean"]
```

```python
with open("metrics.json", "w") as fh:
    json.dump(summary, fh)
```
